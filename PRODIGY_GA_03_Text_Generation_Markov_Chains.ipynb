Creating Training Dataset

```python
training_text = """
A young programmer learns Python and builds simple applications.
The programmer practices coding every day and improves technical skills.
An engineering student studies programming and develops useful projects.
The student learns from mistakes and tests the application carefully.
A developer writes clean code and solves difficult programming problems.
Good programming requires practice patience and logical thinking.
Students work on projects to gain practical engineering experience.
A developer can improve skills by building projects and learning new tools.
Programming helps students understand technology and solve real world problems.
An engineer designs useful systems and tests every part of the project.
"""

with open("markov_training.txt", "w", encoding="utf-8") as file:
    file.write(training_text)

print("Training dataset created successfully!")
print("Number of characters:", len(training_text))
```

Converting Text into Words

```python
import re

# Clean and split the training text into words
words = re.findall(r"\b[\w]+\b", training_text.lower())

print("Text converted into words successfully!")
print("Total words:", len(words))
print("\nFirst 20 words:")
print(words[:20])
```

Creating Markov Chain Model

```python
from collections import defaultdict, Counter

# Create a first-order Markov Chain
markov_chain = defaultdict(Counter)

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    markov_chain[current_word][next_word] += 1

print("Markov Chain created successfully!")
print("Number of states:", len(markov_chain))

# Show some example transitions
print("\nExample transitions:")

for word in list(markov_chain.keys())[:10]:
    print(word, "->", dict(markov_chain[word]))
```

Converting Frequencies into Probabilities

```python
# Convert transition frequencies into probabilities

probability_chain = {}

for current_word, next_words in markov_chain.items():
    total = sum(next_words.values())

    probability_chain[current_word] = {
        word: count / total
        for word, count in next_words.items()
    }

print("Probability model created successfully!")

# Show some example probabilities
print("\nExample probabilities:")

for word in list(probability_chain.keys())[:10]:
    print(word, "->", probability_chain[word])
```

Generating New Text

```python
import random

def generate_text(start_word, length=30):
    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(length - 1):
        if current_word not in probability_chain:
            break

        next_words = list(probability_chain[current_word].keys())
        probabilities = list(probability_chain[current_word].values())

        current_word = random.choices(
            next_words,
            weights=probabilities,
            k=1
        )[0]

        generated_words.append(current_word)

    return " ".join(generated_words)


# Generate text
generated_text = generate_text("a", 30)

print("Generated Text:")
print(generated_text)
```

Generating More Texts

```python
print("Generated Text Samples:\n")

start_words = ["a", "the", "an"]

for i, start_word in enumerate(start_words, 1):
    print(f"Sample {i}:")
    print(generate_text(start_word, 30))
    print()
```

Saving Generated Text

```python
with open("generated_text.txt", "w", encoding="utf-8") as file:
    file.write("Markov Chain Text Generation Results\n")
    file.write("=" * 40 + "\n\n")

    for i, start_word in enumerate(start_words, 1):
        text = generate_text(start_word, 30)
        file.write(f"Sample {i} (Start word: {start_word})\n")
        file.write(text + "\n\n")

print("Generated text saved successfully!")
print("File name: generated_text.txt")
```
